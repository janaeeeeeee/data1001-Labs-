```python
import pandas as pd
import numpy as np

# Step 1: Load the dataset (skipping the title rows)[cite: 1]
df = pd.read_excel('CARICOMS-TOTAL-EXPORTS-BY-COUNTRY-1973-2024-r.xlsx', skiprows=4)

# Step 2: Set row 0 as the column names and drop it from the data[cite: 1]
df.columns = df.iloc[0]
df = df.drop(0).reset_index(drop=True)

# Step 3: Replace text placeholder symbols ("…" and "gr") with NaN[cite: 1]
df = df.replace(["…", "gr"], np.nan)

# Step 4: Convert export columns to numbers (converting any remaining messy text to NaN)[cite: 1]
for column in df.columns:
    if column != 'Year':
        df[column] = pd.to_numeric(df[column], errors='coerce')

# Check missing values count for each column
print("--- Missing Values Per Column ---")
print(df.isna().sum())

# Part B2: Fill all missing export values with 0
df = df.fillna(0)

# Verify that missing values are all gone
print("Missing values after filling:")
print(df.isna().sum().sum())

# B3: Convert Year to an integer
df['Year'] = df['Year'].astype(int)

# B4: Clean up column header text (remove newlines like \n)
df.columns = df.columns.str.replace('\n', ' ').str.strip()

# Checksd the column names and data types to confirm
print("--- Cleaned Column Names ---")
print(df.columns)

print("\n--- Data Types ---")
print(df.dtypes)
```
